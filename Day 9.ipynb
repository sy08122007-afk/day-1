```python
import numpy as np
from sklearn.linear_model import LinearRegression
# Input data: study hours
hours = np.array([[1], [2], [3], [4], [5]])

# Output data: exam scores
scores = np.array([35, 45, 55, 65, 75])

# Create and train the model
model = LinearRegression()
model.fit(hours, scores)

# Predict score for 6 study hours
prediction = model.predict([[6]])

print("Predicted score for 6 hours of study:", prediction[0])

# Display the equation: y = mx + c
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

hours = np.array([[1], [2], [3], [4], [5]])
tickets = np.array([34, 56, 78, 23, 45])

model = LinearRegression()
model.fit(hours, tickets)

prediction = model.predict([[10]])

print("Predicted tickets for 10 hours:", prediction[0])
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
```
