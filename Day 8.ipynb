```python
import pandas as pd
import random

# Sample names
names = [
    "Aarav", "Vivaan", "Aditya", "Arjun", "Rahul",
    "Ananya", "Priya", "Neha", "Kavya", "Sneha"
]

genders = ["Male", "Female"]

employees = []

for i in range(1, 1001):
    employee = {
        "Employee_ID": i,
        "Name": random.choice(names) + " " + str(i),
        "Age": random.randint(21, 60),
        "Gender": random.choice(genders),
        "Salary": random.randint(25000, 150000)
    }
    employees.append(employee)

# Create DataFrame
df = pd.DataFrame(employees)

# Display first 10 rows
print(df.head(10))

# Save as CSV file
df.to_csv("employees_1000.csv", index=False)

print("1000 employee data created successfully!")
```
